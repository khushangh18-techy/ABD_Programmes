```python
import pyspark
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.functions import countDistinct
path = "D:\Khushan_BDA\ABD\SparkSession\Data\indian_food.csv"
```

```python
spark = SparkSession.builder \
    .appName("Indian Food") \
    .getOrCreate()
```

```python
df = spark.read.csv(
    path,
    header=True,
    inferSchema=True
)
```

```python
df.show()
df.printSchema()
```

```python
#unique dishes
df.select(countDistinct("name").alias("Unique_Dishes")).show()
```

```python
#states and dishes
df.groupBy("state") \
    .count() \
    .orderBy(desc("count")) \
    .show()
```

```python
#state with more dishes
df.groupBy("state") \
    .count() \
    .orderBy(desc("count")) \
    .limit(1) \
    .show()
```

```python
#how many from karnataka
df.filter(col("state") == "Karnataka") \
    .count()
```

```python
#List number of unique regions
df.select("state") \
    .distinct() \
    .count()
```

```python
#number of dishes from each region
df.groupBy("state") \
    .count() \
    .orderBy(desc("count")) \
    .show()
```

```python
#unique flavour profiles
df.select("flavor_profile") \
    .distinct() \
    .show()
```

```python
#unique courses
df.select("course") \
    .distinct() \
    .show()
```

```python
#state with more main course
df.filter(col("course") == "main course") \
    .groupBy("state") \
    .count() \
    .orderBy(desc("count")) \
    .limit(1) \
    .show()
```

```python
#% of dishes from each region
total = df.count()

df.groupBy("state") \
    .count() \
    .withColumn(
        "Percentage",
        round(col("count") / total * 100, 2)
    ) \
    .orderBy(desc("Percentage")) \
    .show()
```
